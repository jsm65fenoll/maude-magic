```python
%%maude 
fmod VENDING-MACHINE-SIGNATURE is
    sorts Coin Item Marking .
    subsorts Coin Item < Marking .
    op __ : Marking Marking -> Marking [assoc comm id: null] .
    op null : -> Marking .
    op $ : -> Coin [format (r! o)] .
    op q : -> Item [format (r! o)] .    
    op a : -> Coin [format (b! o)] .        
    op c : -> Coin [format (b! o)] .
endfm        
        
```

```python
%%maude
fmod VENDING-MACHINE-GRAMMAR is
    protecting VENDING-MACHINE-SIGNATURE .
    protecting NAT .

    sort Action .
    op insert $ : -> Action .
    op insert q : -> Action .
    op show basket : -> Action .
    op show credit : -> Action .
    op buy__(s) : Nat Item -> Action .
endfm
        
```
